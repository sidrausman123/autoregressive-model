```python
import random
text = "AI is useful and AI is creative"
words = text.split()

def generate_text(length=12):
    word = random.choice(words)
    result = [word]
    for _ in range(length - 1):
        next_words = [words[i+1] for i in range(len(words) - 1) if words[i] == word]
        word = random.choice(next_words) if next_words else random.choice(words)
        result.append(word)
    return " ".join(result)

print("Generated Text Output")
print(generate_text())
```
